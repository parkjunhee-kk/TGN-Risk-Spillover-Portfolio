```python
import os
import glob
import time
import numpy as np
import pandas as pd
import torch
import networkx as nx
import matplotlib.pyplot as plt
from torch.nn import Linear
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import QuantileRegressor

# PyTorch Geometric
from torch_geometric.nn.models.tgn import TGNMemory, IdentityMessage, LastAggregator
from torch_geometric.nn import TransformerConv

# ==========================================
# 0. 설정 및 장치
# ==========================================
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"🚀 장치 설정: {device}")

# 하이퍼파라미터
memory_dim = 100
time_dim = 100
msg_dim = 2
time_window = 30
rolling_window = 60
num_classes = 12
target_col_idx = 0

# 설정: 임계치 0.001 및 상위 10개 엣지만 시각화
threshold = 0.001
TOP_K_EDGES = 10

investable_assets = ['gold', 'vnq', 'xlk', 'xlf', 'xlv', 'xle', 'xly']

# 테스트 기간 설정
TEST_PERIODS = [
    ('2019-11-01', '2020-05-31'),
    ('2025-01-01', '2025-12-31')
]

# 🎯 이상 탐지 설정
DAYS_CANDIDATES = [7, 14, 21]
ERROR_THRESHOLDS = [2, 3, 4]

# ==========================================
# 1. 모델 정의 및 유틸 함수
# ==========================================
def transfer_entropy(source, target, k=1):
    n = len(source)
    if n <= k: return 0.0
    joint_count = 0
    for i in range(k, n):
        if source[i-1] == target[i]:
            joint_count += 1
    return joint_count / (n - k)

class GraphEmbedding(torch.nn.Module):
    def __init__(self, in_channels, out_channels, msg_dim, time_dim):
        super().__init__()
        self.time_enc = Linear(1, msg_dim)
        self.conv = TransformerConv(
            in_channels=in_channels, out_channels=out_channels // 2,
            heads=2, dropout=0.1, edge_dim=msg_dim
        )
    def forward(self, x, last_update, edge_index, t, msg):
        rel_t = (t - last_update[edge_index[0]]).float()
        rel_t_emb = self.time_enc(rel_t.view(-1, 1))
        edge_attr = msg + rel_t_emb
        out, (edges, alpha) = self.conv(x, edge_index, edge_attr, return_attention_weights=True)
        return out, edges, alpha

class NodePredictor(torch.nn.Module):
    def __init__(self, embedding_dim, out_dim=12):
        super().__init__()
        self.lin1 = Linear(embedding_dim, embedding_dim // 2)
        self.lin2 = Linear(embedding_dim // 2, out_dim)

    def forward(self, target_node):
        h = self.lin1(target_node)
        h = torch.relu(h)
        return self.lin2(h)

# ==========================================
# 2. 데이터 로더 및 전처리
# ==========================================
def load_csv_data(data_folder='./'):
    csv_files = glob.glob(os.path.join(data_folder, '**', '*.csv'), recursive=True)
    raw_data_dict = {}
    for file_path in csv_files:
        asset_name = os.path.splitext(os.path.basename(file_path))[0]
        if asset_name.lower() == 'manifest': continue
        try:
            df = pd.read_csv(file_path)
            df.columns = df.columns.str.strip()
            rename_dict = {col: 'Date' for col in df.columns if col.lower() in ['date', '날짜', 'time']}
            df.rename(columns=rename_dict, inplace=True)
            raw_data_dict[asset_name] = df
        except Exception as e:
            print(f"[{asset_name}] 로드 에러: {e}")
    return raw_data_dict

def preprocess_for_test(raw_data_dict):
    if 'vkospi' in raw_data_dict: del raw_data_dict['vkospi']
    asset_names = list(raw_data_dict.keys())
    processed_data = {}
    raw_returns_history = {}

    for name in asset_names:
        df = raw_data_dict[name].copy()
        df['Date'] = pd.to_datetime(df['Date'])
        df = df.sort_values('Date').set_index('Date')

        target_col = df.columns[target_col_idx]
        if df[target_col].min() >= 0 and df[target_col].mean() > 1.0:
            df['Log_Return'] = np.log(df[target_col] / df[target_col].shift(1))
        else:
            df['Log_Return'] = df[target_col]

        raw_returns_history[name] = df['Log_Return'].copy()

        df['rolling_mean'] = df['Log_Return'].rolling(window=rolling_window).mean()
        df['rolling_std'] = df['Log_Return'].rolling(window=rolling_window).std()
        df['scaled_value'] = (df['Log_Return'] - df['rolling_mean']) / (df['rolling_std'] + 1e-8)
        df = df.dropna()

        raw_class = np.floor(df['scaled_value'])
        df['target_class'] = (raw_class.clip(-6, 5) + 6).astype(int)

        mask = pd.Series(False, index=df.index)
        for start, end in TEST_PERIODS:
            mask = mask | ((df.index >= start) & (df.index <= end))

        processed_data[name] = df[mask]

    all_dates = set()
    for df in processed_data.values():
        all_dates.update(df.index)
    global_calendar = pd.Series(list(all_dates)).sort_values().reset_index(drop=True)

    returns_df = pd.DataFrame(raw_returns_history)
    return global_calendar, processed_data, returns_df, asset_names

# ==========================================
# 3. TENET 분석 및 네트워크 시각화
# ==========================================
def analyze_tenet_risk(df_returns, tau=0.05, alpha_penalty=0.005):
    asset_names = df_returns.columns.tolist()
    scaler = StandardScaler()
    df_scaled = pd.DataFrame(scaler.fit_transform(df_returns), columns=asset_names, index=df_returns.index)
    adj_matrix = pd.DataFrame(0.0, index=asset_names, columns=asset_names)

    for target_asset in asset_names:
        y = df_scaled[target_asset]
        X = df_scaled.drop(columns=[target_asset])
        qr = QuantileRegressor(quantile=tau, alpha=alpha_penalty, solver='highs')
        qr.fit(X, y)
        for i, feature_asset in enumerate(X.columns):
            adj_matrix.loc[feature_asset, target_asset] = np.abs(qr.coef_[i])

    G = nx.DiGraph()
    for emitter in asset_names:
        for receiver in asset_names:
            weight = adj_matrix.loc[emitter, receiver]
            if weight > 0.0 and emitter != receiver:
                G.add_edge(emitter, receiver, weight=weight)
    return G

def plot_focused_networks(date, asset_names, tgn_edges, tgn_attention, tenet_G, event_desc, center_node, top_k=10):
    fig, axes = plt.subplots(1, 2, figsize=(18, 8))
    fig.suptitle(f"🚨 이상 현상 분석: {date.date()} | 기준 노드: [{center_node}]\n({event_desc})", fontsize=18, fontweight='bold')

    MIN_WIDTH = 1.0
    MAX_WIDTH = 10.0

    # --- 1. TGN Attention Network ---
    edges_np = tgn_edges.cpu().numpy()
    if tgn_attention.dim() > 1:
        att_np = tgn_attention.mean(dim=1).cpu().numpy()
    else:
        att_np = tgn_attention.cpu().numpy()

    tgn_edge_list = []
    for i in range(edges_np.shape[1]):
        u, v = asset_names[edges_np[0, i]], asset_names[edges_np[1, i]]
        w = att_np[i]
        if u == center_node or v == center_node:
            tgn_edge_list.append((u, v, w))

    tgn_edge_list = sorted(tgn_edge_list, key=lambda x: x[2], reverse=True)[:top_k]

    max_tgn_w = max([w for _, _, w in tgn_edge_list]) if tgn_edge_list else 1.0

    G_tgn = nx.DiGraph()
    for u, v, w in tgn_edge_list:
        scaled_weight = MIN_WIDTH + (w / max_tgn_w) * (MAX_WIDTH - MIN_WIDTH)
        G_tgn.add_edge(u, v, weight=scaled_weight)

    ax1 = axes[0]
    if len(G_tgn.nodes) > 0:
        pos_tgn = nx.spring_layout(G_tgn, seed=42, k=2.0, iterations=100)
        node_colors = ['orange' if node == center_node else 'lightblue' for node in G_tgn.nodes()]
        edges_tgn = G_tgn.edges()
        weights_tgn = [G_tgn[u][v]['weight'] for u, v in edges_tgn]

        nx.draw_networkx_nodes(G_tgn, pos_tgn, ax=ax1, node_color=node_colors, node_size=2500, edgecolors='black', alpha=0.95)
        nx.draw_networkx_labels(G_tgn, pos_tgn, ax=ax1, font_size=11, font_weight='bold')
        nx.draw_networkx_edges(G_tgn, pos_tgn, ax=ax1, edgelist=edges_tgn, width=weights_tgn, edge_color='blue', alpha=0.6, arrowsize=20, connectionstyle='arc3, rad=0.2')
    ax1.set_title(f"TGN Risk Transfer (Top {top_k} around {center_node})", fontsize=14)
    ax1.axis('off')

    # --- 2. TENET Network ---
    tenet_edge_list = []
    for u, v, data in tenet_G.edges(data=True):
        if u == center_node or v == center_node:
            tenet_edge_list.append((u, v, data['weight']))

    tenet_edge_list = sorted(tenet_edge_list, key=lambda x: x[2], reverse=True)[:top_k]

    max_tenet_w = max([w for _, _, w in tenet_edge_list]) if tenet_edge_list else 1.0

    G_tenet = nx.DiGraph()
    for u, v, w in tenet_edge_list:
        scaled_weight = MIN_WIDTH + (w / max_tenet_w) * (MAX_WIDTH - MIN_WIDTH)
        G_tenet.add_edge(u, v, weight=scaled_weight)

    ax2 = axes[1]
    if len(G_tenet.nodes) > 0:
        pos_tenet = nx.spring_layout(G_tenet, seed=42, k=2.0, iterations=100)
        node_colors_tenet = ['red' if node == center_node else 'lightcoral' for node in G_tenet.nodes()]
        edges_tenet = G_tenet.edges()
        weights_tenet = [G_tenet[u][v]['weight'] for u, v in edges_tenet]

        nx.draw_networkx_nodes(G_tenet, pos_tenet, ax=ax2, node_color=node_colors_tenet, node_size=2500, edgecolors='black', alpha=0.95)
        nx.draw_networkx_labels(G_tenet, pos_tenet, ax=ax2, font_size=11, font_weight='bold')
        nx.draw_networkx_edges(G_tenet, pos_tenet, ax=ax2, edgelist=edges_tenet, width=weights_tenet, edge_color='crimson', alpha=0.6, arrowsize=20, connectionstyle='arc3, rad=0.2')
    ax2.set_title(f"TENET Risk Transfer (Top {top_k} around {center_node})", fontsize=14)
    ax2.axis('off')

    plt.tight_layout()
    plt.show()

# ==========================================
# 4. 메인 테스트 및 이상 탐지 파이프라인
# ==========================================
def run_anomaly_testing():
    data_dir = '/content/drive/MyDrive/Find-A final'
    save_dir = '/content/drive/MyDrive/Find-A final'

    raw_data_dict = load_csv_data(data_dir)
    global_calendar, processed_data, returns_df, asset_names = preprocess_for_test(raw_data_dict)
    num_asset = len(asset_names)
    all_nodes_tensor = torch.arange(num_asset, dtype=torch.long).to(device)

    print(f"✅ 테스트 대상 기간 데이터 포인트 수: {len(global_calendar)}")

    predict_modes = ['core', 'others', 'all']
    bins = [-1.0, 1.0]

    for mode in predict_modes:
        print(f"\n{'='*60}\n🔍 [{mode.upper()}] 모드 모델 로드 및 테스트 시작\n{'='*60}")

        memory = TGNMemory(num_nodes=num_asset, raw_msg_dim=msg_dim, memory_dim=memory_dim, time_dim=time_dim,
                           message_module=IdentityMessage(msg_dim, memory_dim, time_dim), aggregator_module=LastAggregator()).to(device)
        gnn = GraphEmbedding(memory_dim, memory_dim, msg_dim, time_dim).to(device)
        predictor = NodePredictor(embedding_dim=memory_dim, out_dim=num_classes).to(device)

        try:
            memory.load_state_dict(torch.load(os.path.join(save_dir, f'best_memory_{mode}.pth')))
            gnn.load_state_dict(torch.load(os.path.join(save_dir, f'best_gnn_{mode}.pth')))
            predictor.load_state_dict(torch.load(os.path.join(save_dir, f'best_predictor_{mode}.pth')))
        except Exception as e:
            print(f"⚠️ {mode} 모드 가중치 로드 실패: {e}")
            continue

        memory.eval(); gnn.eval(); predictor.eval()

        anomaly_counts = {f"{d}일_오차{e}이상": 0 for d in DAYS_CANDIDATES for e in ERROR_THRESHOLDS}
        extreme_value_counts = 0
        error_history = {name: [] for name in asset_names}

        with torch.no_grad():
            memory.reset_state()

            for t_idx in range(time_window, len(global_calendar) - 1):
                current_date = global_calendar[t_idx]
                next_date = global_calendar[t_idx + 1]

                active_nodes = [i for i, name in enumerate(asset_names) if current_date in processed_data[name].index and not pd.isna(processed_data[name].loc[current_date, 'scaled_value'])]
                if not active_nodes: continue

                edge_sources, edge_targets, edge_features = [], [], []

                for i in active_nodes:
                    name_i = asset_names[i]
                    current_return = processed_data[name_i].loc[current_date, 'scaled_value']

                    edge_sources.append(i); edge_targets.append(i); edge_features.append([current_return, 1.0])
                    data_i = processed_data[name_i].loc[:current_date, 'scaled_value'].tail(time_window).values

                    for j in active_nodes:
                        if i == j: continue
                        name_j = asset_names[j]
                        data_j = processed_data[name_j].loc[:current_date, 'scaled_value'].tail(time_window).values

                        if len(data_i) == time_window and len(data_j) == time_window:
                            source_discrete = np.digitize(data_i, bins)
                            target_discrete = np.digitize(data_j, bins)

                            te_value = transfer_entropy(source_discrete, target_discrete, k=1)
                            if te_value >= threshold:
                                edge_sources.append(i); edge_targets.append(j); edge_features.append([current_return, te_value])

                if not edge_sources: continue

                edge_index = torch.tensor([edge_sources, edge_targets], dtype=torch.long).to(device)
                edge_attr = torch.tensor(edge_features, dtype=torch.float).to(device)

                valid_nodes, targets = [], []
                for node in active_nodes:
                    name = asset_names[node]
                    if next_date in processed_data[name].index:
                        targets.append(processed_data[name].loc[next_date, 'target_class'])
                        valid_nodes.append(node)

                if not valid_nodes: continue

                t_tensor = torch.full((edge_index.size(1),), t_idx, dtype=torch.long).to(device)
                z, last_update = memory(all_nodes_tensor)

                h, attn_edges, attn_weights = gnn(z, last_update, edge_index, t_tensor, edge_attr)
                logits = predictor(h[valid_nodes])
                preds = logits.argmax(dim=1)
                y_target = torch.tensor(targets, dtype=torch.long).to(device)

                memory.update_state(edge_index[0], edge_index[1], t_tensor, edge_attr)

                # ======================================
                # 이상 현상 판별
                # ======================================
                anomaly_triggered = False
                triggered_desc = ""
                anomaly_node = ""

                for idx, node in enumerate(valid_nodes):
                    name = asset_names[node]
                    pred_val = preds[idx].item()
                    true_val = y_target[idx].item()

                    if pred_val in [0, 1, 10, 11]:
                        extreme_value_counts += 1
                        anomaly_triggered = True
                        triggered_desc = f'극단값 예측 (클래스 {pred_val})'
                        anomaly_node = name
                        break

                    current_error = abs(pred_val - true_val)
                    error_history[name].append(current_error)

                    max_d = max(DAYS_CANDIDATES)
                    if len(error_history[name]) > max_d:
                        error_history[name].pop(0)

                    for d in DAYS_CANDIDATES:
                        for e in ERROR_THRESHOLDS:
                            if len(error_history[name]) >= d:
                                recent_errors = error_history[name][-d:]
                                if all(err >= e for err in recent_errors):
                                    rule_key = f"{d}일_오차{e}이상"
                                    anomaly_counts[rule_key] += 1
                                    anomaly_triggered = True
                                    triggered_desc = f'연속 {d}일 동안 오차 {e} 이상 발생'
                                    anomaly_node = name
                                    error_history[name].clear()
                                    break
                        if anomaly_triggered: break
                    if anomaly_triggered: break

                # ======================================
                # 이상 발생 시 시각화
                # ======================================
                if anomaly_triggered:
                    print(f"\n🚨 [Anomaly Detected] 날짜: {current_date.date()} | 발생 노드: {anomaly_node} | 사유: {triggered_desc}")

                    past_60_days = returns_df.loc[:current_date].tail(60)
                    past_60_days = past_60_days.ffill().bfill().dropna(axis=1)

                    if len(past_60_days) > 30 and past_60_days.shape[1] >= 2:
                        tenet_G = analyze_tenet_risk(past_60_days)
                        plot_focused_networks(
                            date=current_date,
                            asset_names=asset_names,
                            tgn_edges=attn_edges,
                            tgn_attention=attn_weights,
                            tenet_G=tenet_G,
                            event_desc=triggered_desc,
                            center_node=anomaly_node,
                            top_k=TOP_K_EDGES
                        )
                    else:
                        print("⚠️ TENET 생성을 위한 유효 데이터 부족으로 네트워크 생성을 건너뜁니다.")

        print(f"\n📊 [{mode.upper()}] 모드 탐지 결과 요약")
        print(f" - 극단값 이상 현상 탐지 횟수: {extreme_value_counts}")
        for rule_key, count in anomaly_counts.items():
            print(f" - {rule_key} 탐지 횟수: {count}")

if __name__ == "__main__":
    run_anomaly_testing()
```
